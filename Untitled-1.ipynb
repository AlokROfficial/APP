```python
def make_multiplier(factor):
    def multiplier(number):
        return number * factor
    return multiplier
double =make_multiplier(2)
triple =make_multiplier(3)
print("double: ", double)
print("triple: ", triple)

print("double: ", double(5))
print("triple: ", triple(5))

print("Inspect the double memory:", double.__closure__[0])
print("Inspect the triple memory:", triple.__closure__[0])
```

```python
def decorator(function):
    def wrapper(a,b):
        print("addition started")
        result=function(a,b)
        print("addition completed")
        return result
    return wrapper
@decorator
def add(a,b):
    return a+b
print(add(10,20)) 
  
```

```python
import time
def timer(function):
    def wrapper(*args,**kwargs):
        start=time.time()
        result=function(*args,**kwargs)
        end=time.time()
        print("Execution Time:", end - start,"seconds")
        return result
    return wrapper
@timer
def calculate_sum():
    total=0
    for i in range(1000000):
        total+=i
    return total
print("result:",calculate_sum())
```

```python
def authenticate(function):
    def wrapper(user,*args,**kwargs):
        if user=="admin":
            return function(user,*args,**kwargs)
        print("Authentication failed")
    return wrapper
def log(function):
    def wrapp(*args, **kwargs):
        print("Transaction started")
        result=function(*args,**kwargs)
        print("Transaction completed")
        return result
    return wrapp
@authenticate
@log     
def transfer_money(user, amount):
    print("Transactional Rs.",amount)
transfer_money("student",5000)
transfer_money("admin",5000)
```

```python
from functools import wraps
def retry(times):
    def decorator(function):
        @wraps (function)
        def wrapp(*args, **kwargs):
            for attempt in range(1,times+1):
                try:
                    return function(*args,**kwargs)
                except Exception as e:
                    print("Attempt",attempt,"failed:",e)
            print("All attempts failed")
        return wrapp
    return decorator
@retry(3)
def connect_server():
    print("connecting to server...")
    raise ConnectionError("Server Unavialable")
connect_server()

```
