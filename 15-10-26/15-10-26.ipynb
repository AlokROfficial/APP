```python
import pickle 
import os
f="s.pkl"
def load_recorder():
    if os.path.exists(f):
        with open(f,"rb") as file:
            return pickle.load(file)
    return []
def save_records(records):
    with open(f,"wb") as file:
        pickle.dump(records, file)
def add_student(records):
    roll_no= input("Enter roll number:").strip()
    if any(s["roll_no"]== roll_no for s in records):
        print("Roll number already exits")
        return
    name=input("Enter name: ").strip()
    marks=float(input("Enter marks: "))
    records.append({"roll_no": roll_no,"name":name,"marks":marks})
    save_records(records)
    print("student record saved")
def display_students(records):
    if not records:
        print("No student records found")
        return
    for s in records:
        print(s)
def search_student(records):
    roll_no= input("Enter roll number to search:").strip()
    for s in records:
        if s["roll_no"]== roll_no:
            print("Student record found:",s)
            return
    print("Student record not found")
records=load_recorder()
while True:
    print("\n1. Add student record")
    print("2. Display all student records")
    print("3. Search student record by roll number")
    print("4. Exit")
    choice=input("Enter your choice:").strip()
    if choice=="1":
        add_student(records)
    elif choice=="2":
        display_students(records)
    elif choice=="3":
        search_student(records)
    elif choice=="4":
        break
    else:
        print("Invalid choice. Please try again.")
```
