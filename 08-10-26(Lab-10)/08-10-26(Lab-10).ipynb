```python
def outer():
    mess="Hello"
    def inner():
        print(mess)
    return inner
greet=outer()
greet()
print("\n Inspect closure:\n",greet)
print("\n Inspect using __closure__:\n",greet.__closure__)
print("\n Cell content:\n",greet.__closure__[0].cell_contents)    
```

```python
def create_account(intial_balance):
    balance=intial_balance
    def deposit(amount):
        nonlocal balance
        balance +=amount
        return balance
    return deposit
account=create_account(5000)
print(account(1000))
print(account(500))
```
